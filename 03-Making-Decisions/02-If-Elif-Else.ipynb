# 🚦 if, elif and else

## 🎯 What you'll learn
- How to make your code **choose** what to do
- Why **indentation** (spaces) matters in Python

---
## 1. `if`: do something only when it's True
Think of a traffic light 🚦: **if** it's green, you go.

````python
light = "green"

if light == "green":
    print("Go! 🚗")
````

### 📏 Indentation is the rule!
The code that belongs to the `if` is pushed in by **4 spaces** (press `Tab`). Python uses these spaces to know which lines are "inside" the `if`.

Don't forget the **colon `:`** at the end of the `if` line!

````python
temperature = 35

if temperature > 30:
    print("It's hot! 🥵")          # inside the if
    print("Drink some water.")    # also inside
print("This line always runs.")   # outside the if
````

## 2. `else`: otherwise...

````python
age = 15

if age >= 18:
    print("You can vote. 🗳️")
else:
    print("You can vote in", 18 - age, "years.")
````

## 3. `elif`: more choices
`elif` means **"else if"**. Python checks the conditions from top to bottom and runs **only the first one that is True**.

````python
score = 82

if score >= 90:
    grade = "A+"
elif score >= 80:
    grade = "A"
elif score >= 70:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
````

## 4. `if` inside `if`

````python
is_logged_in = True
role = "admin"

if is_logged_in:
    print("Welcome!")
    if role == "admin":
        print("You can see the admin panel. 🔧")
else:
    print("Please log in.")
````

## 5. Using truthy/falsy values
Remember from Chapter 01: empty things count as False.

````python
username = ""

if username:
    print(f"Hello, {username}!")
else:
    print("Please enter a username.")
````

## 6. One-line if (shortcut)

````python
stock = 0
status = "In stock" if stock > 0 else "Sold out"
print(status)
````

> ### 🔗 Django Connection
>
> A Django **view** (the code that runs when someone visits a page) is full of `if` statements:
>
> ```python
> def dashboard(request):
>     if not request.user.is_authenticated:
>         return redirect("login")           # not logged in → go to login page
>     if request.method == "POST":
>         # the user submitted a form
>         ...
>     else:
>         # the user is just looking at the page
>         ...
> ```
>
> `if request.method == "POST":` is one of the most common lines in all of Django!

## ✅ Recap
- `if condition:`, then **indented** code runs only when it's True.
