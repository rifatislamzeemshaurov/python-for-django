# ⚖️ Comparisons and Logic

## 🎯 What you'll learn
- How to **compare** values (bigger? equal?)
- How to combine questions with `and`, `or`, `not`

Every comparison is a **question** that Python answers with `True` or `False`.

---
## 1. Comparison operators

| Symbol | Question | Example | Answer |
|--------|----------|---------|--------|
| `==` | equal? | `5 == 5` | `True` |
| `!=` | not equal? | `5 != 3` | `True` |
| `>` | bigger? | `5 > 3` | `True` |
| `<` | smaller? | `5 < 3` | `False` |
| `>=` | bigger or equal? | `5 >= 5` | `True` |
| `<=` | smaller or equal? | `4 <= 3` | `False` |

⚠️ **`=` vs `==`**: one `=` **puts** a value in a box. Two `==` **asks** whether two things are equal.

````python
age = 16
print(age == 16)
print(age != 16)
print(age >= 18)
print(age < 18)
````

````python
# Strings can be compared too (capital letters matter!)
print("python" == "python")   # True
print("Python" == "python")   # False
print("python" == "Python".lower())   # True
````

## 2. Logical operators: combining questions 🧩

| Operator | Meaning | True when... |
|----------|---------|--------------|
| `and` | both | **both** sides are True |
| `or` | either | **at least one** side is True |
| `not` | flip | flips True ↔ False |

🎢 **Real life example:** to ride a roller coaster you must be **at least 10 years old** `and` **at least 120 cm tall**.

````python
age = 12
height = 115

print(age >= 10 and height >= 120)   # False: too short
print(age >= 10 or height >= 120)    # True: one is enough
print(not age >= 10)                 # False
````

````python
is_logged_in = True
is_banned = False

can_post = is_logged_in and not is_banned
print("Can post a comment?", can_post)
````

## 3. Chained comparisons ⛓️
Python lets you check a range in a natural way:

````python
score = 75
print(0 <= score <= 100)    # is the score between 0 and 100?
print(60 <= score < 80)     # is it a B grade?
````

## 4. Membership: `in` ✔️

````python
allowed_roles = ["admin", "editor"]
role = "editor"
print(role in allowed_roles)
````

## ✅ Recap
- `==` asks "equal?"; `=` stores a value.
- `and` means both, `or` means at least one, `not` flips it.
