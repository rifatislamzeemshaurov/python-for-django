## 1. How a website works 🌐
When you visit a website, this happens in less than a second:

```
  👩 You type:  mysite.com/blog/
        │
        ▼   1. REQUEST  ("Please give me /blog/")
 ┌─────────────────────────────────────────────┐
 │                DJANGO SERVER                │
 │                                             │
 │  urls.py   →  "/blog/" goes to post_list    │  ← URL patterns (a list!)
 │     │                                       │
 │     ▼                                       │
 │  views.py  →  def post_list(request): ...   │  ← a FUNCTION
 │     │                                       │
 │     ▼                                       │
 │  models.py →  Post.objects.all()            │  ← a CLASS (talks to the database)
 │     │                                       │
 │     ▼                                       │
 │  template  →  {% for post in posts %} ...   │  ← HTML + a for loop
 └─────────────────────────────────────────────┘
        │
        ▼   2. RESPONSE (an HTML page)
  👩 You see the blog page in your browser!
```

You built **every one of these pieces** in the Mini Blog project. 💪

## 3. ✅ Am I ready? Quick self-check
Run this cell. If you can predict every output **before** running it, you're ready! 😎

````python
from datetime import date

class Post:
    def __init__(self, title, likes=0, tags=None):
        self.title = title
        self.likes = likes
        self.tags = tags or []

    def __str__(self):
        return f"{self.title} ({self.likes} ❤️)"

posts = [
    Post("Hello Django", 12, ["django"]),
    Post("Python Tips", 30, ["python", "tips"]),
    Post("My Setup", 5),
]

context = {
    "today": date(2026, 10, 24).strftime("%d %b %Y"),
    "popular": [str(p) for p in posts if p.likes >= 10],
    "all_tags": sorted({tag for p in posts for tag in p.tags}),
}

for key, value in context.items():
    print(f"{key}: {value}")
````

## 4. 🚀 Your first Django project, step by step
Open a **terminal** (not this notebook) and follow along.

### Step 1: Create a project folder and a virtual environment
```bash
mkdir my_blog
cd my_blog
python -m venv venv
venv\Scripts\activate          # Windows
# source venv/bin/activate     # Mac / Linux
```

### Step 2: Install Django
```bash
pip install django
django-admin --version
```

### Step 3: Create the project
```bash
django-admin startproject mysite .
```
(Don't forget the dot `.` at the end! It means "put it in this folder".)

### Step 4: Run the server 🎉
```bash
python manage.py runserver
```
Open **http://127.0.0.1:8000/** in your browser. You'll see a rocket 🚀. **That's your website!**
Press `Ctrl + C` in the terminal to stop the server.

### Step 5: Create your first app
```bash
python manage.py startapp blog
```
Then add `"blog"` to the `INSTALLED_APPS` **list** in `mysite/settings.py`.

## 5. 👀 A sneak peek at real Django code
Read these slowly. You'll understand every line now!

**`blog/models.py`**: a **class** (Chapter 07)
```python
from django.db import models

class Post(models.Model):
    title = models.CharField(max_length=200)
    content = models.TextField()
    created_at = models.DateTimeField(auto_now_add=True)
    is_published = models.BooleanField(default=False)

    def __str__(self):
        return self.title
```

**`blog/views.py`**: a **function** with a **dictionary** (Chapters 05 and 02)
```python
from django.shortcuts import render, get_object_or_404
from .models import Post

def post_list(request):
    posts = Post.objects.filter(is_published=True)
    return render(request, "blog/post_list.html", {"posts": posts})

def post_detail(request, pk):
    post = get_object_or_404(Post, pk=pk)
    return render(request, "blog/post_detail.html", {"post": post})
```

**`blog/urls.py`**: a **list** (Chapter 02)
```python
from django.urls import path
from . import views

urlpatterns = [
    path("", views.post_list, name="post_list"),
    path("post/<int:pk>/", views.post_detail, name="post_detail"),
]
```

**`blog/templates/blog/post_list.html`**: a **for loop** and an **if** in HTML (Chapters 03 and 04)
```html
<h1>My Blog</h1>
{% for post in posts %}
    <h2><a href="{% url 'post_detail' post.pk %}">{{ post.title }}</a></h2>
    <p>{{ post.created_at|date:"d M Y" }}</p>
{% empty %}
    <p>No posts yet. 😢</p>
{% endfor %}
```
