# 🔄 While Loops, break and continue

## 🎯 What you'll learn
- `while`: keep going **as long as** something is True
- `break`: stop the loop now
- `continue`: skip to the next round

---
## 1. `while` loops
A `while` loop repeats **while** a condition is True. Like: *"while the plate is not empty, keep eating"* 🍛.

````python
count = 1

while count <= 5:
    print("Count is", count)
    count += 1      # ⚠️ don't forget this!

print("Done!")
````

### ⚠️ Watch out for infinite loops!
If the condition never becomes False, the loop runs **forever**. In the example above, without `count += 1`, `count` would stay `1` forever.

😱 If your notebook gets stuck, press the **■ Stop** button (or `I` `I` in command mode) to interrupt it.

## 2. `for` or `while`?
- Use **`for`** when you know **what** to loop over (a list, a range). ✅ Most common!
- Use **`while`** when you **don't know how many times**. You wait for something to happen.

## 3. `break`: stop right now 🛑

````python
numbers = [3, 8, 15, 4, 22, 9]

for n in numbers:
    if n > 10:
        print("Found a big number:", n)
        break        # stop searching
    print("Checking", n)
````

## 4. `continue`: skip this one ⏭️

````python
for n in range(1, 11):
    if n % 2 == 0:
        continue     # skip even numbers
    print(n)
````

## 5. 🎮 Mini Game: Guess the number!
The computer picks a secret number. You have **3 tries**. Run the cell and play!

````python
import random

secret = random.randint(1, 10)   # a random number from 1 to 10
tries = 0

print("I'm thinking of a number between 1 and 10 🤔")

while tries < 3:
    guess = int(input("Your guess: "))
    tries += 1

    if guess == secret:
        print(f"🎉 Correct! You got it in {tries} tries!")
        break
    elif guess < secret:
        print("Too low! ⬇️")
    else:
        print("Too high! ⬆️")
else:
    print(f"😢 Out of tries! The number was {secret}.")
````

💡 Did you notice the `else` after the `while`? It runs only if the loop ended **without** a `break`. It's a neat Python trick!

> ### 🔗 Django Connection
>
> You won't write many `while` loops in Django. Django handles the "waiting for visitors" loop for you!
> When you run `python manage.py runserver`, Django starts a loop that waits for people to visit your website, forever, until you press `Ctrl + C`. 😄
>
> But `break` and `continue` are useful in any `for` loop, for example skipping unpublished posts:
> ```python
> for post in posts:
>     if not post.is_published:
>         continue
>     print(post.title)
> ```

## ✅ Recap
- `while condition:` repeats as long as the condition is True.
- Always change something inside, or you'll get an infinite loop!
- `break` stops the loop; `continue` skips to the next round.
- Prefer `for` loops. They're more common in web development.
