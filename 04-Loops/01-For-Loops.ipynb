# 🔁 For Loops

## 🎯 What you'll learn
- How to repeat code for **every item** in a list
- `range()`, `enumerate()` and `zip()`
- How to loop through dictionaries

---
## 1. Why loops?
Imagine printing 100 student names by hand... 😩 A loop does the boring work for you!

A `for` loop means: **"for each item in this collection, do this."**

````python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("I like", fruit)
````

Read it like English: *"for each **fruit** in **fruits**, print it."*

`fruit` is a temporary variable. Each time around the loop it holds the next item. You can name it anything, but a good habit is to use the singular word: `for post in posts`, `for user in users`.

````python
# Loop through the letters of a string
for letter in "Django":
    print(letter)
````

## 2. `range()`: loop a number of times

| Code | Numbers you get |
|------|-----------------|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |

````python
for i in range(1, 6):
    print("Push-up number", i, "💪")
````

## 3. Adding things up (the "accumulator")
A very common pattern: start at 0 and add each item.

````python
prices = [120, 450, 80, 300]
total = 0

for price in prices:
    total += price

print("Cart total:", total)
````

## 4. `enumerate()`: get the position too

````python
winners = ["Rifat", "Sara", "Tanvir"]

for position, name in enumerate(winners, start=1):
    print(f"#{position}: {name}")
````

## 5. `zip()`: loop through two lists together 🤝

````python
students = ["Rifat", "Sara", "Tanvir"]
marks = [88, 95, 72]

for student, mark in zip(students, marks):
    print(f"{student} got {mark}")
````

## 6. Looping through dictionaries 📖

````python
profile = {"name": "Rifat", "city": "Dhaka", "role": "student"}

for key, value in profile.items():
    print(f"{key}: {value}")
````

## 7. ⭐ Looping through a list of dictionaries
This is what you'll do in **every** Django project:

````python
posts = [
    {"title": "Hello World", "author": "Rifat", "likes": 10},
    {"title": "Learning Python", "author": "Sara", "likes": 25},
    {"title": "Django is Fun", "author": "Rifat", "likes": 42},
]

for post in posts:
    print(f"📝 {post['title']} by {post['author']} ({post['likes']} likes)")
````

````python
# Loop + if = filtering!
print("Posts by Rifat:")
for post in posts:
    if post["author"] == "Rifat":
        print("-", post["title"])
````

> ### 🔗 Django Connection
>
> In a Django HTML **template**, a for loop shows every blog post on the page:
>
> ```html
> <h1>My Blog</h1>
> {% for post in posts %}
>     <h2>{{ post.title }}</h2>
>     <p>by {{ post.author }}</p>
> {% endfor %}
> ```
>
> It's the same idea as the Python loop you just wrote! 🎉

## ✅ Recap
- `for item in collection:` repeats code for every item.
- `range(1, 6)` gives the numbers 1 to 5.
- `enumerate()` gives positions; `zip()` pairs two lists.
- `for key, value in d.items():` loops through a dictionary.
