## 1. What is a module?
A **module** is just a **Python file** (`.py`) full of code you can reuse. Python comes with a huge library of built-in modules. Think of it as a toolbox 🧰.

````python
import math

print(math.sqrt(16))   # square root
print(math.pi)
````

````python
import random

print(random.randint(1, 6))                        # roll a dice 🎲
print(random.choice(["rock", "paper", "scissors"]))
````

## 2. Different ways to import

````python
# 1. import the whole module, then use module.thing
import math
print(math.floor(4.7))

# 2. import only what you need
from math import ceil, pi
print(ceil(4.2), pi)

# 3. give it a nickname
import datetime as dt
print(dt.date.today())
````

⚠️ Avoid `from module import *`. It brings in everything and makes it hard to see where things come from.

## 3. Making your own module 🛠️
Look in this folder: there's a file called **`greetings.py`**. Open it in your editor! It contains:

```python
def say_hello(name):
    return f"Hello, {name}!"

def say_goodbye(name):
    return f"Goodbye, {name}! See you soon."

if __name__ == "__main__":
    print("You ran greetings.py directly!")
    print(say_hello("World"))
```

Because it's in the same folder, we can import it:

````python
import greetings

print(greetings.say_hello("Rifat"))
print(greetings.say_goodbye("Sara"))
````

````python
from greetings import say_hello
print(say_hello("Tanvir"))
````

## 4. `if __name__ == "__main__":` 🤔
Notice that `"You ran greetings.py directly!"` was **not** printed when we imported it. Why?

- When you **run** a file directly (`python greetings.py`), Python sets `__name__` to `"__main__"`.
- When you **import** it, `__name__` is the file's name (`"greetings"`).

So the code under `if __name__ == "__main__":` runs **only when you run the file directly**. It's perfect for test code.

````python
print(greetings.__name__)
````

👉 Try it: open a terminal in this folder and run `python greetings.py`.

## 5. Packages: folders of modules 📁
A **package** is a **folder** of modules with a special file called `__init__.py` inside. This folder has one called **`shop/`**:

```
shop/
├── __init__.py     ← makes "shop" a package (can be empty)
├── products.py
└── cart.py
```

````python
from shop.products import get_price
from shop.cart import Cart

print(get_price("pen"))

my_cart = Cart()
my_cart.add("pen")
my_cart.add("notebook")
print(my_cart.total())
````

## 6. pip: installing packages from the internet 🌍
There are **hundreds of thousands** of free packages on **PyPI** (https://pypi.org). **Django is one of them!**

You install packages with **pip** in your **terminal** (not in Python):

```bash
pip install requests        # install a package
pip install django          # install Django!
pip list                    # see what's installed
pip uninstall requests      # remove a package
```

## 7. 🏠 Virtual environments (venv): very important!
Imagine two projects:
- Project A needs **Django 4**
- Project B needs **Django 5**

If you install everything on your computer directly, they fight! 🥊
A **virtual environment** is a **separate little room** for each project, with its own packages.

**Create and use one (do this for EVERY Django project):**

```bash
# 1. go to your project folder, then create the environment
python -m venv venv

# 2. turn it on (activate)
venv\Scripts\activate          # Windows
source venv/bin/activate       # Mac / Linux

# 3. now install packages into it
pip install django

# 4. turn it off when you're done
deactivate
```

✅ When it's active, you'll see `(venv)` at the start of your terminal line.

## 8. requirements.txt 📝
A list of all the packages your project needs, so others can install them in one go:

```bash
pip freeze > requirements.txt       # save your package list
pip install -r requirements.txt     # install everything from the list
```

This course has its own `requirements.txt` in the main folder. Take a look!

> ### 🔗 Django Connection
>
> **A Django project is a package of packages!** When you create one, you get this:
>
> ```
> mysite/
> ├── manage.py
> ├── mysite/              ← a package (has __init__.py)
> │   ├── __init__.py
> │   ├── settings.py      ← a module
> │   └── urls.py
> └── blog/                ← an "app" = another package
>     ├── __init__.py
>     ├── models.py
>     └── views.py
> ```
>
> And you'll write imports like these every day:
> ```python
> from django.shortcuts import render, redirect
> from django.http import HttpResponse
> from .models import Post          # the dot means "from this same package"
> ```

## ✅ Recap
- A **module** is a `.py` file; a **package** is a folder with `__init__.py`.
