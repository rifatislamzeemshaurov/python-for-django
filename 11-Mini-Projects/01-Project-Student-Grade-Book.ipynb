## Step 2: Calculate an average 🧮
Write a function `average(marks)` that returns the average, rounded to 1 decimal place. If the list is **empty**, return `0` (no dividing by zero!).

````python
# ✍️ Your code here
````

<details>
<summary>👀 <b>Show solution</b> (try it yourself first!)</summary>

```python
def average(marks):
    if not marks:
        return 0
    return round(sum(marks) / len(marks), 1)

print(average([80, 85, 88]))   # 84.3
print(average([]))             # 0
```

```python
def get_grade(avg):
    if avg >= 90:
        return "A+"
    elif avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    return "F"

print(get_grade(91.7), get_grade(84.3), get_grade(58))
```

```python
    for student in students:
        avg = average(student["marks"])
        print(f"📄 {student['name']:<7}| avg: {avg} | grade: {get_grade(avg)}")
```

```python
def add_student(students, name, marks):
    if not name.strip():
        raise ValueError("Name can't be empty")
    for mark in marks:
        if not 0 <= mark <= 100:
            raise ValueError(f"Invalid mark: {mark}")
    students.append({"name": name.strip(), "marks": marks})

for name, marks in [("Mitu", [77, 81, 79]), ("", [50]), ("Adib", [90, 120])]:
    try:
        add_student(students, name, marks)
        print(f"✅ Added {name}")
    except ValueError as error:
        print(f"❌ Could not add {name!r}: {error}")

print_report(students)
```

```python
class GradeBook:
    def __init__(self):
        self.students = []

    def add_student(self, name, marks):
        add_student(self.students, name, marks)

    def report(self):
        print_report(self.students)

    def top_student(self):
        return max(self.students, key=lambda s: average(s["marks"]))["name"]

book = GradeBook()
book.add_student("Rifat", [80, 85, 88])
book.add_student("Sara", [95, 90, 90])
book.report()
print("🏆", book.top_student())
```
