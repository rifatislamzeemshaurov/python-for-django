# 👨‍👧 Inheritance

## 🎯 What you'll learn
- How a class can **inherit** (get for free) everything from another class
- **Overriding** methods
- `super()`: calling the parent's version

⭐ Every Django model and class-based view uses inheritance!

---
## 1. What is inheritance?
Children inherit things from their parents: eye colour, surname... 👪
In Python, a **child class** gets all the attributes and methods of its **parent class**, for free!

```
        Animal           ← parent: has name, eat()
       /      \
     Dog      Cat        ← children: get name and eat() for free,
   bark()    meow()        plus their own extra methods
```

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating 🍖")

class Dog(Animal):        # Dog inherits from Animal
    def bark(self):
        print(f"{self.name} says Woof! 🐶")

class Cat(Animal):
    def meow(self):
        print(f"{self.name} says Meow! 🐱")

rex = Dog("Rex")
rex.eat()     # came from Animal!
rex.bark()    # Dog's own method

kitty = Cat("Kitty")
kitty.eat()
kitty.meow()
````

## 2. Overriding: doing it differently 🔄
A child can **replace** a parent's method by writing a method with the same name.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

class Dog(Animal):
    def speak(self):            # overrides Animal.speak
        return "Woof!"

class Cow(Animal):
    def speak(self):
        return "Moo!"

for animal in [Dog("Rex"), Cow("Bella"), Animal("Mystery")]:
    print(animal.name, "→", animal.speak())
````

## 3. `super()`: use the parent's method too 🤝
Sometimes the child wants to **add** to the parent's work, not replace it. `super()` calls the parent's version.

````python
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

    def info(self):
        return f"{self.username} ({self.email})"

class Admin(User):
    def __init__(self, username, email, level):
        super().__init__(username, email)   # let User set username & email
        self.level = level                   # then add the extra bit

    def info(self):
        return super().info() + f" ⭐ Admin level {self.level}"

normal = User("sara", "sara@mail.com")
boss = Admin("rifat", "rifat@mail.com", 3)

print(normal.info())
print(boss.info())
````

## 4. Checking the family tree 🌳

````python
print(isinstance(boss, Admin))    # True
print(isinstance(boss, User))     # True: an Admin IS a User
print(isinstance(normal, Admin))  # False
````

## 5. 🧪 Let's build a mini "Django-style" model system!
Here's a parent class that gives **every** model a `save()` method and an automatic `id`. Then our models just inherit it.

````python
class Model:
    _next_id = 1

    def save(self):
        if not hasattr(self, "id"):
            self.id = Model._next_id
            Model._next_id += 1
        print(f"💾 Saved {self.__class__.__name__} #{self.id}: {self}")


class Post(Model):
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return self.title


class Comment(Model):
    def __init__(self, text):
        self.text = text

    def __str__(self):
        return self.text


Post("Hello Django").save()
Post("Inheritance is cool").save()
Comment("Great post! 👍").save()
````

`Post` and `Comment` never wrote a `save()` method, but they both have one thanks to inheritance. **That's exactly how Django works!**

> ### 🔗 Django Connection
>
> **Models** inherit from `models.Model`, which gives them `.save()`, `.delete()`, `.objects` and much more, for free:
> ```python
> class Post(models.Model):
>     title = models.CharField(max_length=200)
> ```
>
> **Class-based views** inherit from Django's ready-made views:
> ```python
> from django.views.generic import ListView
>
> class PostListView(ListView):     # inherits all the "show a list" logic
>     model = Post
>     template_name = "blog/post_list.html"
> ```
>
> **Overriding + `super()`**, to add your own logic to `save()`:
> ```python
> class Post(models.Model):
>     ...
>     def save(self, *args, **kwargs):
>         self.slug = self.title.lower().replace(" ", "-")
>         super().save(*args, **kwargs)     # then let Django do the real saving
> ```

class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

for shape in [Rectangle(4, 5), Circle(3)]:
    print(shape.__class__.__name__, shape.area())

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: {self.price} taka"
