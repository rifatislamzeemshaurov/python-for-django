# 🏭 Classes and Objects

## 🎯 What you'll learn
- What a **class** and an **object** are
- `__init__` and `self` (they're less scary than they look! 😄)
- **Attributes** (data) and **methods** (actions)
- `__str__`: how an object describes itself

⭐ **Super important for Django:** every database table in Django (a **model**) is a class!

---
## 1. Blueprint vs the real thing 🍪
A **class** is a **cookie cutter**. An **object** is a **cookie** made with it.

```
   CLASS (cookie cutter)          OBJECTS (cookies)
   ┌────────────────┐         🍪 student1: Rifat, 20
   │    Student     │  ───▶   🍪 student2: Sara, 19
   │ name, age      │         🍪 student3: Tanvir, 22
   └────────────────┘
```

One class → as many objects as you want. Each object has its **own** data.

## 2. Your first class

````python
class Dog:
    pass       # an empty class (pass means "nothing here yet")

dog1 = Dog()   # make an object from the class
dog2 = Dog()
print(dog1)
print(type(dog1))
````

🏷️ Class names use **CapitalWords** (`Dog`, `BlogPost`), not snake_case.

## 3. `__init__` and `self`
`__init__` is a special method that runs **automatically** when you create a new object. It sets up the object's starting data.

`self` means **"this particular object"**: the cookie being made right now.

````python
class Student:
    def __init__(self, name, age):
        self.name = name     # save the name INTO this object
        self.age = age

rifat = Student("Rifat", 20)
sara = Student("Sara", 19)

print(rifat.name, rifat.age)
print(sara.name, sara.age)
````

🧠 **What happened?**
1. `Student("Rifat", 20)` creates a new empty object.
2. Python calls `__init__(self=<the new object>, name="Rifat", age=20)`.
3. `self.name = name` puts `"Rifat"` inside the object.

`self.name` is called an **attribute**: a variable that belongs to an object.

## 4. Methods: things an object can do 🏃
A **method** is a function that lives inside a class. Its first parameter is always `self`.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.courses = []          # every student starts with no courses

    def introduce(self):
        return f"Hi, I'm {self.name} and I'm {self.age}."

    def enroll(self, course):
        self.courses.append(course)
        print(f"{self.name} enrolled in {course} ✅")

rifat = Student("Rifat", 20)
print(rifat.introduce())
rifat.enroll("Python")
rifat.enroll("Django")
print(rifat.courses)
````

## 5. Changing attributes

````python
rifat.age = 21
print(rifat.introduce())
````

## 6. `__str__`: a friendly name for your object 🏷️
Printing an object shows something ugly like `<__main__.Student object at 0x...>`. Add `__str__` to make it nice:

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"📚 {self.title} by {self.author}"

book = Book("The Alchemist", "Paulo Coelho", 208)
print(book)
````

## 7. Class attributes: shared by everyone 👨‍👩‍👧
An attribute written **directly in the class** (not in `__init__`) is shared by **all** objects.

````python
class Product:
    currency = "BDT"     # class attribute: the same for all products

    def __init__(self, name, price):
        self.name = name     # instance attribute: different for each product
        self.price = price

    def __str__(self):
        return f"{self.name}: {self.price} {self.currency}"

    def apply_discount(self, percent):
        self.price = self.price - self.price * percent / 100

phone = Product("Phone", 20000)
bag = Product("Bag", 1500)
print(phone)
print(bag)

phone.apply_discount(10)
print(phone)
````

## 8. A list of objects 📋
Just like a list of dictionaries, but nicer!

````python
products = [Product("Pen", 20), Product("Notebook", 120), Product("Backpack", 1800)]

for p in products:
    print(p)

expensive = [p.name for p in products if p.price > 100]
print("Expensive:", expensive)
````

> ### 🔗 Django Connection
>
> **This is what a Django model looks like.** It's a class!
>
> ```python
> from django.db import models
>
> class Post(models.Model):
>     title = models.CharField(max_length=200)        # class attributes = table columns
>     content = models.TextField()
>     likes = models.IntegerField(default=0)
>     is_published = models.BooleanField(default=False)
>
>     def __str__(self):
>         return self.title                           # just like our Book!
> ```
>
> And you use it just like the objects you made today:
> ```python
> post = Post(title="Hello Django", content="My first post!")
> post.likes += 1
> print(post.title)
> ```
>
> Each **object** = one **row** in the database. 🎉

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money!")
        else:
            self.balance -= amount

    def __str__(self):
        return f"{self.owner}'s balance: {self.balance} taka"

account = BankAccount("Rifat")
account.deposit(1000)
account.withdraw(500)
account.withdraw(2000)
print(account)
```

## ✅ Recap
- A **class** is a blueprint; an **object** is made from it: `rifat = Student("Rifat", 20)`.
- `__init__` sets up a new object. `self` means "this object".
- **Attributes** store data (`self.name`). **Methods** are actions (`def enroll(self, ...)`).
- `__str__` controls what `print(object)` shows.
- Django **models** are classes, and each object is one row in the database. ⭐
