## 🌟 5 golden rules for learning to code

1. **Type the code yourself.** Don't just copy and paste. Your fingers learn too!
2. **Break things on purpose.** Change the code and see what happens.
3. **Errors are your friends.** A red error message is Python *helping* you. Read it slowly.
4. **Do the practice.** Every lesson ends with small exercises. That's where the real learning happens.
5. **A little every day** beats a lot once a week.
