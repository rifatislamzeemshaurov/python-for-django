# 🧃 Functions

## 🎯 What you'll learn
- How to make your own **reusable** blocks of code
- **Parameters** (inputs) and **return** (output)
- Default values and keyword arguments

⭐ **This is one of the most important lessons for Django.** Django **views** are just functions!

---
## 1. What is a function?
A function is like a **juice machine** 🧃:
```
   🍎 🍊  (input)
     ↓
 ┌─────────┐
 │ MACHINE │   ← the function does its work
 └─────────┘
     ↓
   🧃      (output)
```
You build the machine **once**, then use it **as many times as you want**.

You've already used built-in functions: `print()`, `len()`, `sum()`, `input()`. Now let's make our own!

## 2. Making a function with `def`

````python
def say_hello():
    print("Hello! 👋")
    print("Welcome to my website.")
````

Nothing happened? That's right! `def` only **builds** the machine. To use it, **call** it with brackets `()`:

````python
say_hello()
say_hello()
````

## 3. Parameters: giving the function input 📥

````python
def greet(name):
    print(f"Hello, {name}! 👋")

greet("Rifat")
greet("Sara")
````

````python
def introduce(name, age):
    print(f"I am {name} and I am {age} years old.")

introduce("Tanvir", 19)
````

## 4. `return`: giving back a result 📤
`print()` just **shows** something. `return` **hands the result back** so you can save it and use it later.

````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
print(add(10, 20) * 2)   # use the result directly
````

### 🤔 print vs return
- `print` = the machine **shouts** the answer, but you can't keep it.
- `return` = the machine **hands you** the answer, so you can save it in a variable.

⚠️ When Python reaches `return`, the function **stops immediately**.

````python
def check_age(age):
    if age >= 18:
        return "Adult"
    return "Child"      # only reached when age < 18

print(check_age(25))
print(check_age(10))
````

## 5. Default values 🎁
Give a parameter a value to use when the caller doesn't provide one:

````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Rifat"))               # uses the default
print(greet("Rifat", "Assalamu Alaikum"))
````

## 6. Keyword arguments 🏷️
You can name the arguments when calling. Then the order doesn't matter, and the code is easier to read!

````python
def create_user(username, email, is_admin=False):
    return {"username": username, "email": email, "is_admin": is_admin}

user = create_user(email="sara@mail.com", username="sara", is_admin=True)
print(user)
````

## 7. Docstrings: explain your function 📝
Write a short description in triple quotes as the first line:

````python
def make_slug(title):
    """Turn a title like 'Hello World' into 'hello-world'."""
    return title.strip().lower().replace(" ", "-")

print(make_slug("  My Awesome Post "))
help(make_slug)
````

## 8. Scope: what happens inside stays inside 🏠
Variables made **inside** a function only exist inside it.

````python
def make_message():
    secret = "I live inside the function"
    return secret

print(make_message())
# print(secret)   ← this would cause an error: secret doesn't exist out here
````

> ### 🔗 Django Connection
>
> **Every Django page is made by a function** called a **view**. It takes a `request` (information about the visitor) and **returns** a response (the web page):
>
> ```python
> from django.http import HttpResponse
>
> def home(request):
>     return HttpResponse("<h1>Welcome to my website!</h1>")
>
> def about(request):
>     return HttpResponse("<h1>About me</h1>")
> ```
>
> Look! `def`, a parameter and `return`, exactly what you just learned. 🎉

## ✅ Recap
- `def name(parameters):` builds a function. `name()` calls it.
- `return` gives a value back and stops the function.
- Default values: `def f(x, y=10)`. Keyword arguments: `f(y=5, x=1)`.
- Django **views** are functions that take a `request` and return a response.
