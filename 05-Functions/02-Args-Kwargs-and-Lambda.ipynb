# 🎒 *args, **kwargs and lambda

## 🎯 What you'll learn
- `*args`: accept **any number** of values
- `**kwargs`: accept **any number** of named values
- `lambda`: tiny one-line functions

You'll see `*args` and `**kwargs` in Django code all the time, so let's make them friendly! 😊

---
## 1. `*args`: a bag of values 🎒
Sometimes you don't know how many values someone will give you. `*args` collects them all into a **tuple**.

````python
def add_all(*numbers):
    print("I received:", numbers)
    return sum(numbers)

print(add_all(1, 2))
print(add_all(5, 10, 15, 20))
````

(The name `args` is just a habit. The magic is the single star `*`.)

## 2. `**kwargs`: a bag of named values 🏷️
`**kwargs` collects **named** values into a **dictionary**.

````python
def build_profile(**info):
    print("I received:", info)
    for key, value in info.items():
        print(f"  {key}: {value}")

build_profile(name="Rifat", city="Dhaka", job="Developer")
````

## 3. Using them together

````python
def order(customer, *items, **options):
    print("Customer:", customer)
    print("Items:", items)
    print("Options:", options)

order("Sara", "burger", "fries", "cola", delivery=True, note="extra ketchup")
````

The order must always be: **normal params → `*args` → `**kwargs`**.

## 4. Unpacking with `*` and `**` 📦
The stars also work the other way: they **open up** a list or dictionary when you call a function.

````python
def introduce(name, age, city):
    print(f"{name}, {age}, from {city}")

data = {"name": "Tanvir", "age": 22, "city": "Sylhet"}
introduce(**data)      # same as introduce(name="Tanvir", age=22, city="Sylhet")

values = ["Mitu", 19, "Khulna"]
introduce(*values)     # same as introduce("Mitu", 19, "Khulna")
````

## 5. `lambda`: tiny functions ⚡
A `lambda` is a small function written in one line, without a name.

````python
# Normal function
def double(x):
    return x * 2

# Same thing as a lambda
double_lambda = lambda x: x * 2

print(double(5), double_lambda(5))
````

The most common use is **sorting** with a `key`, i.e. telling `sorted()` *what to sort by*:

````python
posts = [
    {"title": "Django Tips", "likes": 42},
    {"title": "Hello World", "likes": 10},
    {"title": "Python Tricks", "likes": 25},
]

most_liked = sorted(posts, key=lambda post: post["likes"], reverse=True)
for post in most_liked:
    print(post["likes"], post["title"])
````

````python
names = ["Zara", "adib", "Mitu", "bashir"]
print(sorted(names))                               # capitals come first 🤨
print(sorted(names, key=lambda n: n.lower()))      # proper A to Z
````

> ### 🔗 Django Connection
>
> **1. URL values arrive as keyword arguments.** For the URL `/post/5/`, Django calls your view like this:
> ```python
> def post_detail(request, pk):     # pk = 5
>     ...
> ```
>
> **2. Database filters use keyword arguments:**
> ```python
> Post.objects.filter(author="Rifat", is_published=True)
>
> filters = {"author": "Rifat", "is_published": True}
> Post.objects.filter(**filters)    # same thing, with unpacking!
> ```
>
> **3. You'll often see `*args, **kwargs` in class methods** (Chapter 07):
> ```python
> def save(self, *args, **kwargs):
>     self.slug = make_slug(self.title)
>     super().save(*args, **kwargs)
> ```

```python
def make_tag(tag, text, **attributes):
    attrs = "".join(f' {key}="{value}"' for key, value in attributes.items())
    return f"<{tag}{attrs}>{text}</{tag}>"

print(make_tag("a", "Click me", href="/home", id="nav"))
print(make_tag("h1", "Welcome"))
```

## ✅ Recap
- `*args` collects extra values into a **tuple**.
- `**kwargs` collects extra named values into a **dictionary**.
- `*list` and `**dict` **unpack** values when calling a function.
- `lambda x: ...` is a tiny function, great for `sorted(key=...)`.
