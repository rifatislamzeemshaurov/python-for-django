## 1. Writing a file ✍️
`open(filename, mode)` opens a file. The **mode** says what you want to do:

| Mode | Meaning |
|------|---------|
| `"r"` | **read** (the default) |
| `"w"` | **write** (⚠️ replaces everything in the file!) |
| `"a"` | **append** (add to the end) |

We always use **`with`**. It closes the file automatically when we're done, even if an error happens. 🧹

````python
with open("my_notes.txt", "w", encoding="utf-8") as file:
    file.write("Day 1: I learned Python variables.\n")
    file.write("Day 2: I learned lists and dictionaries.\n")

print("File saved! ✅ (Look for my_notes.txt in this folder)")
````

💡 `encoding="utf-8"` makes sure emojis and non-English letters (like বাংলা) are saved correctly.

## 2. Reading a file 📖

````python
with open("my_notes.txt", "r", encoding="utf-8") as file:
    content = file.read()      # the whole file as one string

print(content)
````

````python
# Read line by line
with open("my_notes.txt", encoding="utf-8") as file:
    for line in file:
        print("👉", line.strip())   # strip() removes the "\n" at the end
````

## 3. Adding to a file ➕

````python
with open("my_notes.txt", "a", encoding="utf-8") as file:
    file.write("Day 3: I learned files! 📂\n")

with open("my_notes.txt", encoding="utf-8") as file:
    print(file.read())
````

## 4. What if the file doesn't exist? 🤷

````python
try:
    with open("does_not_exist.txt") as file:
        print(file.read())
except FileNotFoundError:
    print("😅 That file doesn't exist!")
````

## 5. 🌐 JSON: how websites talk
**JSON** (JavaScript Object Notation) is a text format for sharing data between programs. When an app on your phone gets weather data, news or messages from a server, it almost always arrives as JSON.

Good news: **JSON looks almost exactly like Python dictionaries and lists!** 🎉

```json
{
    "name": "Rifat",
    "age": 20,
    "is_student": true,
    "skills": ["Python", "Django"]
}
```

Small differences: JSON uses `true`/`false`/`null` instead of `True`/`False`/`None`, and **always double quotes**.

Python's built-in `json` module translates for us:

| Function | Direction |
|----------|-----------|
| `json.dumps(data)` | Python → JSON **text** |
| `json.loads(text)` | JSON **text** → Python |
| `json.dump(data, file)` | Python → JSON **file** |
| `json.load(file)` | JSON **file** → Python |

💡 Remember: the **s** stands for **string**.

````python
import json

user = {"name": "Rifat", "age": 20, "is_student": True, "skills": ["Python", "Django"], "phone": None}

json_text = json.dumps(user, indent=2)
print(json_text)
print(type(json_text))   # it's a string now
````

````python
# JSON text coming from a website...
response = '{"city": "Dhaka", "temperature": 31, "sunny": true}'

weather = json.loads(response)   # ...becomes a Python dictionary
print(weather["city"], weather["temperature"])
print(type(weather))
````

### Saving and loading JSON files 💾

````python
posts = [
    {"id": 1, "title": "Hello World", "likes": 10},
    {"id": 2, "title": "Learning JSON", "likes": 7},
]

with open("posts.json", "w", encoding="utf-8") as file:
    json.dump(posts, file, indent=2)

with open("posts.json", encoding="utf-8") as file:
    loaded = json.load(file)

for post in loaded:
    print(post["title"])
````

### 🧹 Clean up
Let's delete the practice files we made:

````python
import os

for name in ["my_notes.txt", "posts.json"]:
    if os.path.exists(name):
        os.remove(name)
print("Cleaned up! 🧹")
````

> ### 🔗 Django Connection
>
> - Django stores your data in a **database**, not in text files, so you'll rarely use `open()` for that. You'll still use files for **uploads** (profile pictures, PDFs).
> - **JSON is huge in Django.** Building an API (data for mobile apps or JavaScript) looks like this:
>   ```python
>   from django.http import JsonResponse
>
>   def api_posts(request):
>       data = [{"id": p.id, "title": p.title} for p in Post.objects.all()]
>       return JsonResponse(data, safe=False)
>   ```
> - **Django REST Framework**, the most popular way to build APIs, is all about JSON.

## ✅ Recap
- Always use `with open(name, mode, encoding="utf-8") as file:`.
- `"w"` overwrites, `"a"` adds, `"r"` reads.
- JSON is how websites share data, and it looks like dicts and lists.
- `json.dumps` / `json.loads` work with strings; `json.dump` / `json.load` work with files.
