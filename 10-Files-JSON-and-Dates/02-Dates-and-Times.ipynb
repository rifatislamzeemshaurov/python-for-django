# 📅 Dates and Times

## 🎯 What you'll learn
- How to get today's date and the current time
- How to do **math with dates** (like "7 days from now")
- How to **format** dates nicely: `24 Oct 2026`

Every blog post has a "published on" date. Every message has a timestamp. Every order has a delivery date. Dates are everywhere on the web! 🗓️

---
## 1. The `datetime` module

````python
from datetime import date, datetime, timedelta

today = date.today()
now = datetime.now()

print("Today:", today)
print("Now:", now)
````

````python
print(today.year, today.month, today.day)
print(now.hour, now.minute)
````

## 2. Making your own dates

````python
birthday = date(2005, 12, 16)                 # year, month, day
meeting = datetime(2026, 10, 24, 14, 30)      # + hour, minute
print(birthday)
print(meeting)
````

## 3. Date math with `timedelta` ➕➖
A `timedelta` is an **amount of time**: 7 days, 3 hours, and so on.

````python
today = date.today()

next_week = today + timedelta(days=7)
last_month = today - timedelta(days=30)

print("Next week:", next_week)
print("30 days ago:", last_month)
````

````python
# How many days between two dates?
new_year = date(today.year + 1, 1, 1)
days_left = (new_year - today).days
print(f"🎆 {days_left} days until New Year!")
````

````python
# Free trial ends in 14 days
signup = datetime.now()
trial_ends = signup + timedelta(days=14)
print("Your free trial ends on", trial_ends.date())
````

## 4. Formatting dates nicely with `strftime` 🎨
`strftime` turns a date into text in **any format you like** (think: **str**ing **f**rom **time**).

| Code | Meaning | Example |
|------|---------|---------|
| `%d` | day | `24` |
| `%m` | month number | `10` |
| `%B` | month name | `October` |
| `%b` | short month | `Oct` |
| `%Y` | year | `2026` |
| `%A` | weekday | `Saturday` |
| `%H:%M` | 24-hour time | `14:30` |
| `%I:%M %p` | 12-hour time | `02:30 PM` |

````python
meeting = datetime(2026, 10, 24, 14, 30)

print(meeting.strftime("%d/%m/%Y"))            # 24/10/2026
print(meeting.strftime("%d %B %Y"))            # 24 October 2026
print(meeting.strftime("%A, %b %d"))           # Saturday, Oct 24
print(meeting.strftime("%I:%M %p"))            # 02:30 PM
````

## 5. Text → date with `strptime` 🔄
The opposite direction: turn text (like from a form) into a real date (**str**ing **p**arse **time**).

````python
text = "16-12-2005"
dob = datetime.strptime(text, "%d-%m-%Y")
print(dob)
print(type(dob))
````

## 6. 🧪 "Posted 3 days ago"
A classic website feature!

````python
def time_ago(posted):
    seconds = (datetime.now() - posted).total_seconds()
    if seconds < 60:
        return "just now"
    if seconds < 3600:
        return f"{int(seconds // 60)} minutes ago"
    if seconds < 86400:
        return f"{int(seconds // 3600)} hours ago"
    return f"{int(seconds // 86400)} days ago"

print(time_ago(datetime.now() - timedelta(seconds=20)))
print(time_ago(datetime.now() - timedelta(minutes=45)))
print(time_ago(datetime.now() - timedelta(hours=5)))
print(time_ago(datetime.now() - timedelta(days=3)))
````

> ### 🔗 Django Connection
>
> Django fills dates in for you:
> ```python
> class Post(models.Model):
>     created_at = models.DateTimeField(auto_now_add=True)   # set once, when created
>     updated_at = models.DateTimeField(auto_now=True)       # updated on every save
> ```
>
> In Django, use `timezone.now()` instead of `datetime.now()` (it knows about time zones):
> ```python
> from django.utils import timezone
> week_ago = timezone.now() - timedelta(days=7)
> recent_posts = Post.objects.filter(created_at__gte=week_ago)
> ```
>
> In templates, there are built-in date filters:
> ```html
> {{ post.created_at|date:"d M Y" }}       → 24 Oct 2026
> {{ post.created_at|timesince }} ago      → 3 days ago
> ```
