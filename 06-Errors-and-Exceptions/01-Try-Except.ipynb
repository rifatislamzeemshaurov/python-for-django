# 🚑 Errors and Exceptions

- How to **catch** errors with `try` / `except`
- How to **raise** your own errors

## 1. Errors are your friends 🤝
When something goes wrong, Python stops and shows a red message. It tells you:
1. **Where** the problem is (the line number)
2. **What type** of error it is
3. **A hint** about why

```
Traceback (most recent call last):
  File "example.py", line 3, in <module>
    print(nmae)
NameError: name 'nmae' is not defined     ← read the LAST line first!
```

💡 **Tip:** always read the **last line** of an error first.

## 2. The common errors 🐛

| Error | What it means | Example |
|-------|---------------|---------|
| `SyntaxError` | you typed something Python can't understand | missing `:` or `)` |
| `IndentationError` | wrong spaces at the start of a line | |
| `NameError` | that variable doesn't exist (typo?) | `print(nmae)` |
| `TypeError` | wrong type for this operation | `"5" + 5` |
| `ValueError` | right type, wrong value | `int("hello")` |
| `KeyError` | that key isn't in the dictionary | `user["phone"]` |
| `IndexError` | that position isn't in the list | `[1, 2][10]` |
| `ZeroDivisionError` | dividing by zero | `10 / 0` |
| `AttributeError` | that thing has no such method | `"hi".push()` |

## 3. Catching errors with `try` / `except` 🥅
Put risky code inside `try`. If an error happens, Python jumps to `except` **instead of crashing**.

````python
try:
    number = int("hello")
except ValueError:
    print("😅 That's not a number!")

print("The program keeps running!")
````

### Catching different errors

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("❌ You can't divide by zero!")
        return None
    except TypeError:
        print("❌ Please give me numbers.")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "two"))
````

### See the error message with `as`

````python
user = {"name": "Rifat"}

try:
    print(user["phone"])
except KeyError as error:
    print("Missing key:", error)
````

## 4. `else` and `finally`

| Block | When it runs |
|-------|--------------|
| `try` | always tries first |
| `except` | only if there **was** an error |
| `else` | only if there was **no** error |
| `finally` | **always**, no matter what (good for cleaning up) |

````python
def convert(text):
    try:
        value = int(text)
    except ValueError:
        print(f"'{text}' is not a number ❌")
    else:
        print(f"Converted {text} → {value} ✅")
    finally:
        print("Finished checking.\n")

convert("42")
convert("abc")
````

## 5. Raising your own errors 🚨
Use `raise` when **you** detect a problem:

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age can't be negative!")
    return age

try:
    set_age(-5)
except ValueError as error:
    print("Error:", error)
````

## 6. Making your own error type ✨

````python
class OutOfStockError(Exception):
    pass

def buy(product, stock):
    if stock == 0:
        raise OutOfStockError(f"Sorry, {product} is sold out!")
    return f"You bought a {product} 🛍️"

try:
    print(buy("Laptop", 0))
except OutOfStockError as error:
    print(error)
````

⚠️ **Don't catch everything blindly!** Avoid a bare `except:` that hides every error. Catch the **specific** error you expect.

> ### 🔗 Django Connection
>
> Django uses exceptions a lot. When you look for something in the database that doesn't exist:
>
> ```python
> try:
>     post = Post.objects.get(id=99)
> except Post.DoesNotExist:
>     raise Http404("Post not found")     # shows the "404 Not Found" page
> ```
>
> Django even has a shortcut for this exact pattern:
> ```python
> post = get_object_or_404(Post, id=99)
> ```
>
> Form validation also raises errors:
> ```python
> raise ValidationError("This username is already taken.")
> ```

## ✅ Recap
- Read the **last line** of an error first.
- `try` / `except SomeError:` catches errors so your program doesn't crash.
- `else` runs on success, `finally` always runs.
- `raise ValueError("message")` creates your own error.
- Django: `Post.DoesNotExist`, `Http404`, `get_object_or_404`.
