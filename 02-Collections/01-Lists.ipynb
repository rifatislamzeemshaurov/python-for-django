# 📋 Lists

## 🎯 What you'll learn
- How to keep **many values** in one variable
- How to add, remove and change items
- Sorting and checking what's inside

---
## 1. What is a list?
A list is like a **shopping list** or a **train with carriages** 🚂: many items, in order, inside square brackets `[ ]`.

````python
fruits = ["apple", "banana", "mango"]
scores = [90, 75, 88, 100]
mixed = ["Rifat", 21, True]     # a list can hold different types
empty = []                     # an empty list

print(fruits)
print(len(fruits))   # how many items? → 3
````

## 2. Getting items: indexing
Lists work like strings: positions start at **0**.

```
fruits = [ "apple", "banana", "mango" ]
              0         1         2
             -3        -2        -1
```

````python
fruits = ["apple", "banana", "mango"]
print(fruits[0])    # apple
print(fruits[-1])   # mango (the last one)
print(fruits[0:2])  # ['apple', 'banana'] (slicing works too!)
````

## 3. Changing items
Unlike strings, lists **can be changed**.

````python
fruits = ["apple", "banana", "mango"]
fruits[1] = "orange"
print(fruits)
````

## 4. Adding items ➕

| Method | What it does |
|--------|--------------|
| `.append(x)` | add `x` to the **end** |
| `.insert(i, x)` | add `x` at position `i` |
| `.extend(other)` | add every item from another list |

````python
todo = ["wake up", "brush teeth"]
todo.append("learn Python")
todo.insert(0, "drink water")
todo.extend(["eat breakfast", "go to class"])
print(todo)
````

## 5. Removing items ➖

| Method | What it does |
|--------|--------------|
| `.remove(x)` | remove the first `x` |
| `.pop()` | remove the last item and give it back to you |
| `.pop(i)` | remove the item at position `i` |
| `del list[i]` | delete the item at position `i` |

````python
colors = ["red", "green", "blue", "yellow"]
colors.remove("green")
print(colors)

last = colors.pop()
print("Popped:", last)
print(colors)
````

## 6. Is it in the list? 🔍

````python
guests = ["Rifat", "Sara", "Tanvir"]
print("Sara" in guests)       # True
print("Karim" in guests)      # False
print("Karim" not in guests)  # True
````

## 7. Sorting and other helpers

````python
numbers = [42, 7, 19, 3, 88]

print(sorted(numbers))                # a NEW sorted list (original stays the same)
print(sorted(numbers, reverse=True))  # biggest first
print(sum(numbers), max(numbers), min(numbers))

numbers.sort()                        # sorts the list itself
print(numbers)

names = ["zara", "adib", "mitu"]
names.sort()
print(names)                          # alphabetical order
````

## 8. Lists inside lists
A list can hold other lists, like a table with rows:

````python
classroom = [
    ["Rifat", 90],
    ["Sara", 85],
    ["Tanvir", 78],
]
print(classroom[0])      # ['Rifat', 90]
print(classroom[0][0])   # Rifat
print(classroom[1][1])   # 85
````

> ### 🔗 Django Connection

## ✅ Recap
- Lists hold many items in order: `[1, 2, 3]`.
- Index from 0: `my_list[0]`. The last item is `my_list[-1]`.
- Add with `.append()` / `.insert()`, remove with `.remove()` / `.pop()`.
- `in` checks if something is inside. `sorted()` sorts.
