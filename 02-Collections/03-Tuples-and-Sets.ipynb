# 🔒 Tuples and Sets

## 🎯 What you'll learn
- **Tuples**: lists that can't be changed
- **Unpacking**: opening a tuple into variables
- **Sets**: collections with **no duplicates**

---
## 1. Tuples: locked lists 🔒
A tuple is like a list, but uses round brackets `( )` and **cannot be changed** after you make it. Use it for things that should stay fixed, like a date of birth or the GPS location of a city.

````python
dhaka = (23.81, 90.41)     # latitude, longitude
colors = ("red", "green", "blue")

print(dhaka[0])
print(len(colors))
print("red" in colors)
````

````python
# Trying to change a tuple causes an error:
try:
    colors[0] = "pink"
except TypeError as error:
    print("Oops!", error)
````

(Don't worry about `try`/`except` yet. It just catches the error so the notebook keeps running. You'll learn it in Chapter 06.)

## 2. Unpacking: open the box 🎁
You can pull the values out of a tuple into separate variables in one line:

````python
point = (10, 20)
x, y = point
print(x)
print(y)
````

````python
student = ("Rifat", 20, "Dhaka")
name, age, city = student
print(f"{name} is {age} and lives in {city}")
````

## 3. Sets: no duplicates allowed 🚫👯
A set uses curly brackets `{ }` (like a dictionary, but **without keys**). It automatically removes duplicates and doesn't keep any order.

````python
tags = {"python", "django", "python", "web", "django"}
print(tags)   # duplicates are gone!
````

````python
# A super common trick: remove duplicates from a list
visitors = ["rifat", "sara", "rifat", "tanvir", "sara"]
unique_visitors = set(visitors)
print(unique_visitors)
print(len(unique_visitors), "unique visitors")
````

````python
skills = set()       # an empty set (NOT {}, that's an empty dictionary!)
skills.add("HTML")
skills.add("CSS")
skills.add("HTML")   # already there, nothing happens
print(skills)
````

### Comparing sets

````python
rifat_skills = {"python", "html", "css"}
sara_skills = {"python", "django", "sql"}

print(rifat_skills & sara_skills)   # in BOTH → {'python'}
print(rifat_skills | sara_skills)   # in EITHER (all together)
print(rifat_skills - sara_skills)   # Rifat has, Sara doesn't
````

## 🧠 Which one should I use?

| Type | Brackets | Ordered? | Changeable? | Duplicates? | Use it for... |
|------|----------|----------|-------------|-------------|---------------|
| list | `[ ]` | ✅ | ✅ | ✅ | most collections |
| tuple | `( )` | ✅ | ❌ | ✅ | fixed data |
| set | `{ }` | ❌ | ✅ | ❌ | unique items |
| dict | `{k: v}` | ✅ | ✅ | keys are unique | labelled data |

> ### 🔗 Django Connection
>
> Django uses **tuples** to create dropdown menus (called **choices**):
>
> ```python
> STATUS_CHOICES = [
>     ("draft", "Draft"),          # (value saved in the database, label people see)
>     ("published", "Published"),
> ]
> status = models.CharField(max_length=10, choices=STATUS_CHOICES)
> ```
>
> Each choice is a tuple of two things. Now you can read that! 😎
