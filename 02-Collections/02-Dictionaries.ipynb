# 📖 Dictionaries (The Most Important One for Web!)

## 🎯 What you'll learn
- How to store data as **key → value** pairs
- How to read, add, change and delete data
- Lists of dictionaries, the shape of almost all web data ⭐

---
## 1. What is a dictionary?
Think of a real dictionary 📕: you look up a **word** (the key) and find its **meaning** (the value).

Or think of a **student ID card**:
```
┌──────────────────────────┐
│ name:   Rifat            │
│ age:    20               │
│ class:  Web Development  │
└──────────────────────────┘
```

In Python we write it with curly brackets `{ }`:

````python
student = {
    "name": "Rifat",
    "age": 20,
    "course": "Web Development",
}
print(student)
````

Each line is a **key: value** pair. Keys are usually strings, and values can be anything.

## 2. Reading values

````python
print(student["name"])     # Rifat
print(student["course"])   # Web Development
````

⚠️ If the key doesn't exist, `student["phone"]` crashes with a `KeyError`. The safe way is **`.get()`**:

````python
print(student.get("phone"))                 # None (no crash!)
print(student.get("phone", "Not given"))    # a default value
````

## 3. Adding, changing and deleting

````python
student = {"name": "Rifat", "age": 20}

student["email"] = "rifat@example.com"   # add a new key
student["age"] = 21                      # change an existing key
print(student)

del student["email"]                     # delete a key
print(student)

student.update({"city": "Dhaka", "age": 22})   # add or change many at once
print(student)
````

## 4. Looking inside 🔍

````python
product = {"name": "Laptop", "price": 65000, "in_stock": True}

print(product.keys())     # all the keys
print(product.values())   # all the values
print(product.items())    # all the pairs
print("price" in product) # is this KEY inside? → True
print(len(product))       # how many pairs → 3
````

## 5. Dictionaries inside dictionaries

````python
user = {
    "username": "rifat_dev",
    "profile": {
        "bio": "I love Python!",
        "followers": 120,
    },
}
print(user["profile"]["bio"])
print(user["profile"]["followers"])
````

## 6. ⭐ Lists of dictionaries
This is the **most common shape of data on the web**. A list of blog posts, a list of products, a list of users...

````python
posts = [
    {"id": 1, "title": "Hello World", "author": "Rifat", "likes": 10},
    {"id": 2, "title": "Learning Python", "author": "Sara", "likes": 25},
    {"id": 3, "title": "Django is Fun", "author": "Rifat", "likes": 42},
]

print(len(posts), "posts")
print(posts[0]["title"])      # the first post's title
print(posts[-1]["likes"])     # the last post's likes
````

If you understand this cell, you understand how websites store data. 🎉

> ### 🔗 Django Connection
>
> Dictionaries are **everywhere** in Django!
>
> **1. Sending data to a web page (the "context"):**
> ```python
> def home(request):
>     context = {"name": "Rifat", "posts": posts}
>     return render(request, "home.html", context)
> ```
> Then in the HTML template you can write `{{ name }}`.
>
> **2. Reading form data:** when someone submits a form, Django gives you something dictionary-like:
> ```python
> username = request.POST["username"]
> page = request.GET.get("page", 1)      # .get() with a default, just like we learned!
> ```
>
> **3. JSON / APIs:** JSON data (Chapter 10) looks *exactly* like a Python dictionary.

## ✅ Recap
- Dictionaries store **key: value** pairs: `{"name": "Rifat"}`.
- Read with `d["key"]` or, more safely, `d.get("key", default)`.
- Add/change with `d["key"] = value`, delete with `del d["key"]`.
- `.keys()`, `.values()` and `.items()` let you look inside.
- **Lists of dictionaries** are how web data is shaped. ⭐
