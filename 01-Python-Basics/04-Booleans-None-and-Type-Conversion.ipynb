# ✅ Booleans, None and Type Conversion

## 🎯 What you'll learn
- **Booleans**: `True` and `False`
- **None**: Python's "nothing here" value
- How to **convert** between types (text ↔ numbers)
- How to ask the user for input with `input()`

---
## 1. Booleans: yes or no ✅❌
A boolean can only be **`True`** or **`False`** (with a capital letter!). Think of a light switch: ON or OFF.

````python
is_logged_in = True
is_admin = False
print(is_logged_in)
print(type(is_admin))
````

We usually get booleans by **asking a question** (more about this in Chapter 03):

````python
age = 15
print(age >= 18)   # Is age at least 18? → False
print(age < 18)    # → True
````

## 2. `None`: an empty box 📭
`None` means **"no value yet"**. It is not zero and not empty text. It means *nothing at all*.

````python
middle_name = None
profile_picture = None
print(middle_name)
print(type(middle_name))
````

To check for `None`, use **`is`**:

````python
print(profile_picture is None)       # True
print(profile_picture is not None)   # False
````

## 3. Type conversion: changing a value's type 🔄

| Function | Turns it into... | Example | Result |
|----------|------------------|---------|--------|
| `int()` | a whole number | `int("42")` | `42` |
| `float()` | a decimal number | `float("3.5")` | `3.5` |
| `str()` | text | `str(100)` | `"100"` |
| `bool()` | True / False | `bool(0)` | `False` |

````python
page = "3"                  # text (like from a URL: ?page=3)
page_number = int(page)     # now it's a number
print(page_number + 1)      # 4

points = 95
message = "You scored " + str(points) + " points"   # str() so we can glue
print(message)
````

⚠️ You can only convert text that *looks* like a number. `int("hello")` causes an error. (We'll learn to handle errors in Chapter 06!)

## 4. Truthy and falsy 🤔
Every value counts as either "True-ish" or "False-ish". These are **falsy** (they count as False):

`False`, `None`, `0`, `""` (empty text), `[]` (empty list), `{}` (empty dictionary)

**Everything else** is truthy.

````python
print(bool(""))        # False: empty text
print(bool("Rifat"))   # True
print(bool(0))         # False
print(bool(42))        # True
print(bool([]))        # False: empty list
````

This is handy: `if username:` means *"if the user typed something"*. You'll see this a lot!

## 5. `input()`: asking the user ⌨️
`input()` pauses and waits for the user to type something. **It always gives back a string.**

Run this cell and type your name in the box that appears:

````python
your_name = input("What is your name? ")
print(f"Nice to meet you, {your_name}!")
````

````python
birth_year = input("What year were you born? ")
age = 2026 - int(birth_year)    # convert the text to a number first!
print(f"You are about {age} years old.")
````

💡 In real websites we don't use `input()`. Users type into **forms** on a web page instead. The idea is the same though: **the data arrives as text.**

> ### 🔗 Django Connection
>
> - Django models use booleans all the time:
>   ```python
>   is_published = models.BooleanField(default=False)
>   ```
> - `None` shows up when a field is allowed to be empty:
>   ```python
>   bio = models.TextField(null=True, blank=True)   # bio can be None
>   ```
> - Data from a URL like `/blog/?page=2` arrives as **text**, so we convert it:
>   ```python
>   page = int(request.GET.get("page", 1))
>   ```

## ✅ Recap
- Booleans are `True` or `False`.
- `None` means "no value". Check it with `is None`.
- `int()`, `float()`, `str()` and `bool()` convert between types.
