# 🔤 Strings (Working with Text)

## 🎯 What you'll learn
- What a **string** is
- How to grab letters with **indexing** and **slicing**
- Useful string **methods** like `.upper()` and `.replace()`
- **f-strings**, the easiest way to build sentences ⭐

Websites are made of **text**: titles, usernames, emails, passwords, blog posts. So strings are one of the most important things to learn!

---
## 1. Making strings
A string is text inside quotes. Single `'...'` or double `"..."` quotes both work.

````python
greeting = "Hello"
language = 'Python'
sentence = "It's a sunny day"   # double quotes let you use ' inside
print(greeting, language)
print(sentence)
````

For text with many lines, use **three** quotes:

````python
poem = """Roses are red,
Python is cool,
I'm building websites,
After school!"""
print(poem)
````

## 2. Joining strings and measuring them

````python
first = "Rifat"
last = "Sourov"
full_name = first + " " + last   # + glues strings together
print(full_name)

print(len(full_name))   # len() counts the characters (spaces count too!)
print("Ha" * 3)          # repeat a string
````

## 3. Indexing: grab one letter
Every letter has a **position number** called an **index**. ⚠️ Python starts counting at **0**!

```
 P   y   t   h   o   n
 0   1   2   3   4   5     ← counting from the start
-6  -5  -4  -3  -2  -1     ← counting from the end
```

````python
word = "Python"
print(word[0])    # P (first letter)
print(word[3])    # h
print(word[-1])   # n (last letter)
````

## 4. Slicing: grab a piece
`word[start:stop]` gives you letters from `start` up to, **but not including**, `stop`.

````python
word = "Python"
print(word[0:3])   # Pyt
print(word[2:])    # thon  (from 2 to the end)
print(word[:2])    # Py    (from the start to 2)
print(word[::-1])  # nohtyP (backwards! fun trick)
````

## 5. String methods: built-in superpowers 🦸
A **method** is a tool that belongs to a value. You use it with a dot: `text.method()`.

````python
title = "  learn python for django  "

print(title.upper())        # ALL CAPS
print(title.lower())        # all small
print(title.title())        # Every Word Capitalised
print(title.strip())        # remove spaces from both ends
print(title.replace("python", "PYTHON"))
````

Strings **never change themselves**. A method gives you a *new* string. To keep it, save it:

````python
username = "  RIFAT  "
username = username.strip().lower()   # you can chain methods!
print(username)
````

### More super useful methods

````python
email = "rifat@example.com"

print("@" in email)                  # True: is "@" inside?
print(email.startswith("rifat"))     # True
print(email.endswith(".com"))        # True
print(email.split("@"))              # ['rifat', 'example.com']: cut into pieces
print(email.count("e"))              # how many "e"s?
````

````python
words = ["learn", "python", "today"]
print(" ".join(words))   # glue a list of words with spaces
print("-".join(words))   # or with dashes
````

## 6. ⭐ f-strings: the best way to build text
Put an **`f`** before the quotes and write variables inside **`{curly brackets}`**. Python fills them in for you.

````python
name = "Sourov"
age = 21
print(f"My name is {name} and I am {age} years old.")
````

You can even do math or call methods inside the `{}`:

````python
price = 1250
qty = 2
print(f"Total: {price * qty} taka")
print(f"Shouting: {name.upper()}!")
print(f"Pi is about {3.14159:.2f}")   # :.2f = show 2 decimal places
````

💡 You will use f-strings **all the time** in Django. They are much easier than gluing with `+`.

## 7. Special characters
| Code | Meaning |
|------|---------|
| `\n` | new line |
| `\t` | tab (big space) |
| `\"` | a quote inside quotes |

````python
print("Line one\nLine two")
print("Name:\tRifat")
````

> ### 🔗 Django Connection
>
> In Django every blog post needs a **slug**, a URL-friendly version of its title:
>
> ```
> Title: "My First Django App"   →   URL: mysite.com/blog/my-first-django-app/
> ```
>
> You can already build one with string methods!
>
> ```python
> title = "My First Django App"
> slug = title.lower().replace(" ", "-")   # 'my-first-django-app'
> ```
>
> And in Django models we use f-strings to describe objects:
> ```python
> def __str__(self):
>     return f"{self.title} by {self.author}"
> ```

## ✅ Recap
- Strings are text in quotes. `len()` counts the characters.
- Indexing starts at **0**: `word[0]`. Slicing: `word[1:4]`.
