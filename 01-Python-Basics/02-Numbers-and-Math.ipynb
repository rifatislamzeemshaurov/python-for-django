# 🔢 Numbers and Math

## 🎯 What you'll learn
- The two main kinds of numbers: `int` and `float`
- How to do math in Python
- Handy shortcuts like `+=`

---
## 1. Two kinds of numbers

| Type | Meaning | Examples |
|------|---------|----------|
| `int` | whole numbers | `5`, `-3`, `1000` |
| `float` | numbers with a dot | `3.14`, `0.5`, `-2.0` |

````python
apples = 5        # int
price = 12.50     # float
print(type(apples), type(price))
````

## 2. Python is a super calculator 🧮

| Symbol | Meaning | Example | Result |
|--------|---------|---------|--------|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide | `7 / 2` | `3.5` |
| `//` | divide, drop the decimals | `7 // 2` | `3` |
| `%` | remainder ("what's left over") | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

````python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
````

### 🍕 Understanding `//` and `%` with pizza
You have **7 slices** of pizza and **2 friends**.
- `7 // 2` → each friend gets **3** slices.
- `7 % 2` → **1** slice is left over (you eat it 😋).

A popular trick: `number % 2 == 0` tells you if a number is **even**.

````python
print(10 % 2)   # 0 → 10 is even
print(11 % 2)   # 1 → 11 is odd
````

## 3. Order of operations
Python follows the same rules as school math: brackets first, then `**`, then `* /`, then `+ -`.

````python
print(2 + 3 * 4)      # 14, multiply happens first
print((2 + 3) * 4)    # 20, brackets happen first
````

## 4. Math with variables

````python
price = 250
quantity = 3
total = price * quantity
print("Total:", total, "taka")
````

### ➕ The `+=` shortcut
Adding to a variable is so common that Python has a shortcut:

| Long way | Shortcut |
|----------|----------|
| `x = x + 5` | `x += 5` |
| `x = x - 5` | `x -= 5` |
| `x = x * 5` | `x *= 5` |

````python
likes = 0
likes += 1   # someone liked your post!
likes += 1   # another like!
print("Likes:", likes)
````

## 5. Helpful number tools

````python
print(round(3.14159, 2))   # round to 2 decimal places → 3.14
print(abs(-7))             # distance from zero → 7
print(max(4, 9, 2))        # biggest → 9
print(min(4, 9, 2))        # smallest → 2
````

## ⚠️ Text that looks like a number is still text!
`"5"` (in quotes) is **text**, not a number. Watch what happens:

````python
print("5" + "5")           # joins the text → 55
print(5 + 5)               # adds the numbers → 10
print(int("5") + int("5")) # turn text into numbers first → 10
````

This is **super important** for web development: everything a user types into a website form arrives as **text**. You will often need `int()` to turn it into a number.

> ### 🔗 Django Connection
>
> **Pagination** means splitting a long list into pages ("Page 1 of 5"). Django has a tool for it, but behind the scenes it's simple math:
>
> ```python
> total_posts = 47
> posts_per_page = 10
> total_pages = (total_posts + posts_per_page - 1) // posts_per_page   # 5 pages
> ```
>
> Shopping carts, prices, discounts, star ratings... all just math. 🛒
