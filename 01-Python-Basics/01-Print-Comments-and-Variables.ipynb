# 📦 Print, Comments and Variables

## 🎯 What you'll learn
- How to show things on the screen with `print()`
- How to write notes for humans with **comments**
- How to save information in **variables**

---
## 🖨️ 1. `print()` makes Python talk

`print()` shows whatever you put inside the brackets.

````python
print("Hello, world!")
print("Python is fun")
print(2026)
````

You can print several things at once by separating them with commas. Python puts a space between them automatically.

````python
print("I am", 12, "years old")
````

## 💬 2. Comments: notes for humans

Python **ignores** everything after a `#`. We use comments to explain our code to other people (and to our future selves!).

````python
# This line is a comment. Python skips it.
print("This line runs!")  # comments can also go at the end of a line
````

## 📦 3. Variables: labelled boxes

Imagine a **box with a label on it**. You put something inside and write a name on the box so you can find it later. That's a **variable**!

```
   ┌────────────┐
   │  "Rifat"   │   ← what's inside (the value)
   └────────────┘
       name         ← the label (the variable name)
```

We create a variable with the **`=`** sign. Read it as *"put this value into this box"*.

````python
name = "Rifat"
age = 20
city = "Dhaka"

print(name)
print(age)
print(city)
````

Now we can use the box name instead of typing the value again and again:

````python
print("Hi, my name is", name)
print(name, "lives in", city)
````

### 🔁 Boxes can change
You can put a new value into the same box. The old value is thrown away.

````python
score = 10
print("Score:", score)

score = 25   # new value!
print("Score:", score)
````

### 🏷️ Rules for naming variables

| ✅ Allowed | ❌ Not allowed | Why |
|-----------|----------------|-----|
| `user_name` | `user name` | no spaces |
| `age2` | `2age` | can't start with a number |
| `total_price` | `total-price` | no dashes |
| `first_name` | `class` | some words belong to Python |

🐍 Python people write names in **snake_case**: all lowercase, words joined with `_`.
`blog_post_title` ✅ `BlogPostTitle` ❌ (that style is for *classes*, which we'll meet later)

Python is **case-sensitive**: `Name` and `name` are two different boxes!

### 🔍 What type is inside the box?
Use `type()` to see what kind of value a variable holds.

````python
print(type(name))   # str   = text (a "string")
print(type(age))    # int   = whole number (an "integer")
print(type(3.5))    # float = number with a decimal point
print(type(True))   # bool  = True or False
````

### ✨ Bonus: many boxes in one line

````python
x, y, z = 1, 2, 3
print(x, y, z)
````

> ### 🔗 Django Connection
